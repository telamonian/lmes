```python
%pylab inline
%load_ext autoreload
%autoreload 2

from copy import deepcopy
import os,sys
from pathlib import Path

# numpy.set_printoptions(precision=3, threshold=1e6, linewidth=1e6)

from lm_anal.src.main import Sim, Sims
from lm_anal.src.helper import timewith

def Regen(sim, dataName, excludedFields=None):
    with timewith('regen_%s' % dataName) as t:
        excludedFields = [] if excludedFields is None else excludedFields
        for dataToExcludeFrom,field in excludedFields:
            sim.__getattribute__(dataToExcludeFrom).setExcludedFields(field)
        return sim.__getattribute__(dataName).regen()
```
